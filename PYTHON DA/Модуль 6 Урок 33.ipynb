```python
def sum_numbers(a, b):
    return a + b

print(sum_numbers(3, 5))
```

```python
def is_even(num):
    if num % 2 == 0:
        return True
    else:
        return False

print(is_even(2))
```

```python
def find_max(numbers):
    print(max(numbers))

numbers = [45, 90, 4, 1, 26, 0]
find_max(numbers)
```

```python
def celsius_to_fahrenheit(C):
    F = C * 1.8 + 32
    return F

print(celsius_to_fahrenheit(40))
```

```python
text = "Привет мир"

def count_vowels(text):
    vowels = "аеиоуыэюяАЕИОУЫЭЮЯ"
    count = 0

    for letter in text:
        if letter in vowels:
            count += 1
    
    return count

print(count_vowels(text))
```
