```python
guests = ["Настя", "Катя", "Ангелина"]
print(f"Приглашаю тебя, {guests[0]}, на обед")
print(f"Приглашаю тебя, {guests[1]}, на обед")
print(f"Приглашаю тебя, {guests[2]}, на обед")
```

```python
guests.insert(0, "Руслан")
guests.insert(2, "Артём")
guests.append("Влад")

print("Расширяем список гостей")
print(f"Приглашаю тебя, {guests[0]}, на обед")
print(f"Приглашаю тебя, {guests[1]}, на обед")
print(f"Приглашаю тебя, {guests[2]}, на обед")
print(f"Приглашаю тебя, {guests[3]}, на обед")
print(f"Приглашаю тебя, {guests[4]}, на обед")
print(f"Приглашаю тебя, {guests[5]}, на обед")
```

```python
popped_1_guest = guests.pop(0)
popped_2_guest = guests.pop(1)
popped_3_guest = guests.pop(1)
popped_4_guest = guests.pop()

print("Уменьшаем список гостей")
print(f"Сожалею об отмене приглашения, {popped_1_guest}, на обед")
print(f"Сожалею об отмене приглашения, {popped_2_guest}, на обед")
print(f"Сожалею об отмене приглашения, {popped_3_guest}, на обед")
print(f"Сожалею об отмене приглашения, {popped_4_guest}, на обед")
```
