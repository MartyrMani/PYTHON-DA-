```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def distance_to_origin(self):
        return (self.x ** 2 + self.y ** 2) ** 0.5

point = Point(3, 4)
print(point.distance_to_origin())
```

```python
class Book:
    def __init__(self, title, author, pages):
        self.title = title
        self.author = author
        self.pages = pages

    def is_long(self):
        return self.pages > 500

book_1 = Book("Как завоевывать друзей и оказывать влияние на людей", "Дейл Карнеги", 370)
book_2 = Book("Словарь Жестов", "Жозеф Мессинжер", 703)

print(book_1.is_long())
print(book_2.is_long())
```

```python
class Student:
    def __init__(self, name):
        self.name = name
        self.grades = []

    def add_grade(self, grade):
        self.grades.append(grade)

    def get_average(self):
        return sum(self.grades) / len(self.grades)

student_1 = Student("Толик")
student_2 = Student("Света")

student_1.add_grade(4)
student_1.add_grade(3)
student_1.add_grade(5)
student_1.add_grade(3)
student_1.add_grade(5)

student_2.add_grade(3)
student_2.add_grade(5)
student_2.add_grade(5)
student_2.add_grade(4)
student_2.add_grade(5)

print(student_1.get_average())
print(student_2.get_average())
```
