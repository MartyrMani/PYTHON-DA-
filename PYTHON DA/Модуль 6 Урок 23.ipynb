```python
sum = 0

for i in range(1, 101):
    sum += i

print(sum)
```

```python
num = int(input("Выберите число от 1 до 10: "))

for i in range(1, 11):
    print(f"{num} * {i} = ", num * i)
```

```python
even_num = []

for i in range(2, 21):
    if i % 2 == 0:
        even_num.append(i)

print(even_num)
```
