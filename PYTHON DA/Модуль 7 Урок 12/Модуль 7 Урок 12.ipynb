```python
import pandas as pd

df = pd.read_csv('ДЗ_Pandas.csv')
```

```python
print(df.isnull().sum())
```

```python
df1 = df.dropna(subset = ['Бюджет (млн $)'])
print('Без пропущенного Бюджета')
print(df1)
print()

df.fillna({'Режиссер': 'Неизвестен'}, inplace = True)

df.fillna({'Рейтинг': df['Рейтинг'].mean()}, inplace = True)

df.dropna(subset = ['Бюджет (млн $)', 'Сборы (млн $)'], how = 'all', inplace = True)
print('Очищенная таблица')
print(df)
```

```python
genre = df.groupby('Жанр')[['Рейтинг', 'Бюджет (млн $)']].mean()
print('Группировка по Жанру')
print(genre)
print()

year = df.groupby('Год выпуска').agg(Количество_фильмов=('Название фильма','count'), Суммарные_сборы=('Сборы (млн $)', 'sum'))
print('Группировка по Году Выпуска')
print(year)
print()

combo = df.groupby(['Жанр', 'Год выпуска'])[['Рейтинг']].mean()
print('Комбинированная Группировка')
print(combo)
```

```python
genres = df.groupby('Жанр')['Рейтинг'].mean().reset_index().sort_values('Рейтинг', ascending = False)
print('Группировка и Сортировка')
print(genres)
print()

years = df['Год выпуска'].value_counts().sort_index()
print('Анализ частоты')
print(years)
print()

director_genre = df.groupby(['Режиссер', 'Жанр'])['Сборы (млн $)'].sum().reset_index().sort_values('Сборы (млн $)', ascending = False)
print('Комбинированный Анализ')
print(director_genre)
```
