```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('coffee_sales_data.csv')
print(df)
```

```python
order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
#Для правильного порядка дней недели

sales_by_week = df.groupby('День недели')['Количество Продаж'].sum()
sales_by_week = sales_by_week.reindex(order)

plt.figure(figsize = (10, 6))
plt.bar(sales_by_week.index, sales_by_week.values)
plt.title('Анализ общего количества продаж по дням недели')
plt.xlabel('Дни недели')
plt.ylabel('Продажи')
plt.show()

#Больше всего продаж наблюдается в Субботу, Понедельник и Воскресенье
```

```python
plt.figure(figsize = (10, 6))
df.boxplot(column=['Цена за Чашку'], by=['Тип Кофе'])
plt.title('Анализ цен на разные типы кофе')
plt.xlabel('Тип кофе')
plt.ylabel('Цены')
plt.show()

#В основном они не сильно отличаются (если смотреть по усикам), но всё же видно что больше и длинее всех это Эспрессо
```

```python
day_type = {'Monday': 'Будний', 'Tuesday': 'Будний', 'Wednesday': 'Будний', 'Thursday': 'Будний',
            'Friday': 'Будний', 'Saturday': 'Выходной', 'Sunday': 'Выходной'}
df['Тип дня'] = df['День недели'].map(day_type)

sales_by_type = df.groupby('Тип дня')['Количество Продаж'].sum()

plt.figure(figsize = (10, 6))
plt.bar(sales_by_type.index, sales_by_type.values)
plt.title('Анализ взаимосвязи между выходными и продажами')
plt.xlabel('Тип дня')
plt.ylabel('Продажи')
plt.show()

#В будние дни больше всего количество продаж когда в выходные даже меньше половины от будних
```
