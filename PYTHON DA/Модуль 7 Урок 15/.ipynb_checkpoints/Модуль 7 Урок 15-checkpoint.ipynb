```python
import pandas as pd
df = pd.read_csv('CSE_student_performances.csv')
```

```python
print(df.shape)
print()
print(df.head())
print()
print(df.info())
print()
print(df.describe())
```

```python
df.drop('LikePresentation', axis = 1, inplace = True)
print(df.head())
```

```python
columns = []
columns = df.columns
print(columns)

df = df.rename(columns={'Age ': 'Age', 'Gender': 'Gender', 'AcademicPerformance': 'Academic Performance', 'TakingNoteInClass': 'Taking Note In Class',
       'DepressionStatus': 'Depression Status', 'FaceChallangesToCompleteAcademicTask': 'Face Challanges To Complete Academic Task',
       'SleepPerDayHours': 'Sleep Per Day Hours', 'NumberOfFriend': 'Number Of Friend', 'LikeNewThings': 'Like New Things'})
print(df.head())
```

```python
df.drop_duplicates(inplace = True)
df.dropna(inplace = True)
print(df.shape)
```

```python
print(df['Sleep Per Day Hours'].mean())
```

```python
excellents = df[df['Academic Performance'] == 'Excellent']

print(excellents)
print()
print(excellents['Sleep Per Day Hours'].mean())
```

```python
depressions = df[df['Depression Status'] == 'Yes']

print(depressions)
print()
print(depressions['Sleep Per Day Hours'].mean())
```
