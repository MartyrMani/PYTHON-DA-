```python
class Users:
    def __init__(self, name, surname, birth_year, email):
        self.name = name
        self.surname = surname
        self.birth_year = birth_year
        self.email = email

    def getAge(self):
        return 2023 - self.birth_year

try:
    users = []
    with open('users.txt', 'r') as file:
        for line in file.readlines():
            print(line + "\n")
            parts = line.strip().split(",")
            user = Users(parts[1], parts[2], int(parts[4]), parts[3])
            users.append(user)

    def filter(users, low = 30, high = 50):
        for user in users:
            if low < user.getAge() < high:
                print(f"{user.email}, {user.getAge()}")

    filter(users)
except FileNotFoundError:
    print("The file that you would like to read is not found")
except:
    print("Some error occured")
```
