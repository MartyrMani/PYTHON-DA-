```python
height = int(input("Введите высоту прямоуголника: "))
width = int(input("Введите ширину прямоуголника: "))

for i in range(height):
    for j in range(width):
        print("*", end="")
    print()
```

```python
num1 = int(input("Сколько чисел вы хотите возводить в степень: "))
num2 = int(input("Сколько степеней вы желаете видеть для чисел: "))

for i in range(1, num1 + 1):
    for j in range(1, num2 + 1):
        if j == num2:
            print(f"{i} ^ {j} = ", i ** j, end="")
        else:
            print(f"{i} ^ {j} = ", i ** j, end=", ")
    print()
```

```python
seconds = int(input("Введите секунды для обратного отсчета: "))

while seconds > 0:
    print(seconds)
    seconds -= 1

print("Время истекло!")
```

```python
import random

secret_num = random.randint(1, 100)
print("Угадайте число от 1 до 100!")
num = int(input("Ваше число: "))

while num != secret_num:
    if num < secret_num:
        print("Загаданное число больше!")

    if num > secret_num:
        print("Загаданное число меньше!")

    num = int(input("Ваше число: "))

print("Вы угадали число!")
```
