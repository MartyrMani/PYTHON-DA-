```python
import matplotlib.pyplot as plt
```

```python
days = [30, 31, 1, 2, 3, 4, 5]
temperature = [29, 31, 34, 30, 29, 25, 23]

plt.plot(days, temperature)
plt.title('Температура в городе за неделю')
plt.xlabel('Дни недели')
plt.ylabel('Температура ℃')
plt.show()
```

```python
hours = [21, 15, 31, 34, 28]
grades = [5, 4, 5, 3, 4]

plt.scatter(hours, grades)
plt.title('Взаимосвязь часов с оценками')
plt.xlabel('Часы учёбы')
plt.ylabel('Оценки студентов')
plt.show()
```

```python
sports = ['Футбол', 'Волейбол', 'Баскетбол', 'Лёгкая атлетика', 'Настольный теннис']
count_of_participants = [1063000, 766000, 562000, 384000, 242000]

plt.bar(sports, count_of_participants)
plt.title('Топ-5 видов спорта и количество их участников в Казахстане')
plt.xlabel('Категории спорта')
plt.ylabel('Количество участников')
plt.show()
```

```python
import numpy as np

grades = np.random.randint(2, 6, 25)

plt.hist(grades, bins = 6)
plt.title('Анализ частоты встречаемости различных оценок')
plt.xlabel('Оценки')
plt.ylabel('Количество студентов')
plt.show()
```

```python
coffee = ['Espresso', 'Latte', 'Cappuccino', 'Americano', 'Mocha']
cups = [32, 48, 41, 27, 19]

snacks = ['Croissant', 'Bagel', 'Cookie', 'Muffin', 'Cheesecake']
sold = [24, 18, 35, 21, 12]

plt.subplot(1, 2, 1)
plt.bar(coffee, cups)
plt.title('Количество проданных чашек кофе')
plt.xlabel('Кофе')
plt.ylabel('Количество')

plt.subplot(1, 2, 2)
plt.bar(snacks, sold)
plt.title('Количество проданных закусок')
plt.xlabel('Закуски')
plt.ylabel('Количество')

plt.show()
```
